# Fill Missing Value

```python
df_loan.describe()
```

```python
df_loan.isna().sum()
```

```python
df_loan['Gender'].fillna(df_loan['Gender'].mode()[0], inplace = True)
df_loan['Married'].fillna(df_loan['Married'].mode()[0], inplace = True)
df_loan['Self_Employed'].fillna(df_loan['Self_Employed'].mode()[0], inplace = True)
df_loan['Dependents'].fillna(df_loan['Dependents'].mode()[0], inplace = True)
df_loan['LoanAmount'].fillna(df_loan['LoanAmount'].median(), inplace = True)
df_loan['Loan_Amount_Term'].fillna(df_loan['Loan_Amount_Term'].median(), inplace = True)
df_loan['Credit_History'].fillna(df_loan['Credit_History'].median(), inplace = True)
df_loan.drop(columns = ["Loan_ID"])
```

# To check the Best Model Using GridserachCV

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {

    'n_estimators': [50, 100, 200],

    'max_depth': [None, 5, 10],

    'min_samples_split': [2, 5],

    'min_samples_leaf': [1, 2]

}
```

```python
rf = RandomForestClassifier(random_state=42)
grid_search = GridSearchCV(
    estimator=rf,
    param_grid=param_grid,
    cv=5
)
```

```python
grid_search.fit(X_train,y_train)
```

```python
print(grid_search.best_score_)
```

```python
best_rf= grid_search.best_estimator_
best_rf
```

```python
best_pred = best_rf.predict(X_test)
best_rf
```

```python
print(classification_report(y_test, best_pred))
print(confusion_matrix(y_test, best_pred))
print(accuracy_score(y_test, best_pred))
```
